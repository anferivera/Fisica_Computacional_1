# Computational Fluid Dynamics with FDM

## Navier Stokes's equation

$$\boxed{\rho\bigg(\dfrac{\partial \vec{u}}{\partial t}+ (\vec{u}\cdot \vec{\nabla})\vec{u}\bigg)=-\nabla\vec{p}+\rho\nu\nabla^2\vec{u}+f_{ext}}$$

1. Inercia / Aceleración: El cambio de velocidad del fluido en el tiempo y el espacio.
2. Gradiente de presión $(-\nabla p)$: La fuerza empujada por las diferencias de presión.}
3. Fuerzas viscosas $(\mu \nabla^2 \mathbf{u})$: La resistencia interna del fluido a fluir (fricción).
4. Fuerzas externas $(\mathbf{f})$: Factores externos como la gravedad

Simplify

$$∂u/∂t + (u ⋅ ∇) u = − 1/ρ ∇p + ν ∇²u + \bar{f}$$

u :  Velocity (2d vector)<br>
p :  Pressure<br>
f :  Forcing (here =0)<br>
ν :  Kinematic Viscosity<br>
ρ :  Density<br>

Incompressibility:  ∇ ⋅ u = 0
