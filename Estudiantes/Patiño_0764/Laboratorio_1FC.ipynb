```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = sns.load_dataset("penguins").dropna().values
df = pd.DataFrame(data, columns=["especie", "isla", "longitud_pico (mm)", "profundidad_pico (mm)", "longitud_aleta (mm)", "masa_corporal (g)", "genero"])

df.head()
```

```python
#¿Cuáles son las profundidades del pico máxima y mínima de los pinguinos de la isla Torgersen?
targents = []
for i in range(len(df)):
  if df["isla"][i] == "Torgersen":
    targents.append(df["profundidad_pico (mm)"][i])

targents = np.array(targents)
print(f"Este es el mínimo en la isla de Targents: {targents.min()}")
print(f"Este es el máximo en la isla de Targents: {targents.max()}")
```

```python
#¿De qué isla, especie y género es el pingüino con la mayor masa corporal? Ayuda: utilice la función argmax()
gordito = df["masa_corporal (g)"].argmax()
print(f"posicion del pinguino: {gordito}")
print(df.iloc[gordito])
```

```python
#¿Cuál es la masa corporal media de los pinguinos macho que no son de la especie Gentoo?
masa = df[(df["genero"] == "Male") & (df["especie"] != "Gentoo")]["masa_corporal (g)"]

print(f"La masa corporal media es: { masa.mean():.2f} g")
```

```python
#Grafique la distribución de la longitud del pico. Añada el valor medio y la desviación estándar en el título de la figura. Ayuda: utilice la funcion hist()
plt.figure(figsize=(8,6))

media = df["longitud_pico (mm)"].mean()
desviacion = df["longitud_pico (mm)"].std()

plt.axvline(media, linestyle="--", label=f"Media = {media:.2f} mm")
plt.axvline(desviacion, linestyle="--", label=f"Media = {desviacion:.2f} mm")

plt.hist(df["longitud_pico (mm)"], bins=20)
plt.title("Distribución de la longitud del pico")
plt.xlabel("Longitud del pico")
plt.ylabel("Frecuencia")
plt.legend()
plt.grid()
plt.show()
```

```python
#¿Cuáles son las islas y especies presentes en el conjunto de datos?

print(f"islas del conjunto de datos: {df["isla"].unique()}")
print(f"islas del conjunto de datos: {df["especie"].unique()}")
```

```python
#¿Con qué frecuencia aparecen? Grafique la distribución de la masa corporal para los pingüinos.
plt.figure(figsize=(8,6))

plt.hist(df["masa_corporal (g)"], bins=20)
plt.title("Distribución de la masa corporal")
plt.xlabel("masa corporal")
plt.ylabel("Frecuencia")
plt.grid()
plt.show()
```
