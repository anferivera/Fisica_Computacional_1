````python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


data = sns.load_dataset("penguins").dropna().values
df = pd.DataFrame(data, columns=["especie", "isla", "longitud_pico (mm)", "profundidad_pico (mm)", "longitud_aleta (mm)", "masa_corporal (g)", "genero"])

df[:5]
````

Esto cargará un dataframe (df), usando la librería seaborn, con algunas características morfológicas y geográficas de diferentes especies de pingüinos. Las columnas contienen los siguientes datos:

```
[especie, isla, longitud del pico, profundidad del pico, longitud de la aleta, masa corporal, genero]
```

3.   ¿Cuál es la masa corporal media de los pinguinos macho que no son de la especie Gentoo?

4.   Grafique la distribución de la longitud del pico. Añada el valor medio y la desviación estándar en el título de la figura. **Ayuda:** utilice la funcion `hist()`

5.   ¿Cuáles son las islas y especies presentes en el conjunto de datos? ¿Con qué frecuencia aparecen? Grafique la distribución de la masa corporal para los pingüinos.




  



````python
#1)
prof_pico = df['profundidad_pico (mm)']
pico_max = prof_pico.max()
pico_min = prof_pico.min()
print(f'Pico máximo: {pico_max}')
print(f'Pico mínimo: {pico_min}')
````

````python
#2)
max_masa = df['masa_corporal (g)'].max()
max_masa_fila = df[df['masa_corporal (g)'] == max_masa]
print(f'Los datos del pinguino con mayor masa corporal son:', f'Isla: {max_masa_fila.iloc[0, 1]}, Especie: {max_masa_fila.iloc[0, 0]}, Género: {max_masa_fila.iloc[0, 6]}')
````

````python
#3)
macho_no_gentoo = df[(df['genero'] == 'Male') & (df['especie'] != 'Gentoo')]
media_masa_mng = macho_no_gentoo['masa_corporal (g)'].mean()
print(f'La masa media de los pinguinos machos que no son de la especie Gentoo es: {media_masa_mng} ')
````

````python
#4)
long_pico = df['longitud_pico (mm)']
valor_medio = long_pico.mean()
desv_estandar = long_pico.std()

leyenda = f"Valor medio: {valor_medio:.1f} \nDesviación estándar: {desv_estandar}"


plt.figure()
plt.hist(long_pico, bins=40, edgecolor="black", label= leyenda)
plt.title('Histograma de la longitud del pico')
plt.legend()

````

````python
#5)
isla_columna = df['isla']
especie_columna = df['especie']

islas = isla_columna.unique()
especies = especie_columna.unique()

print(f'Las islas son: {islas}', f'Las especies son: {especies}')
````

````python
frecuencia_islas = isla_columna.value_counts()
frecuencia_islas
````

````python
frecuencia_especies = especie_columna.value_counts()
frecuencia_especies
````

````python
masa = df['masa_corporal (g)']

plt.figure()
plt.hist(masa, bins=40, edgecolor="black", label= 'Masas corporales (g)')
plt.title('Masa corporal')
plt.legend()
````
