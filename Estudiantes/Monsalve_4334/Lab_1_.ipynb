```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = sns.load_dataset("penguins").dropna().values
df = pd.DataFrame(data, columns=["especie", "isla", "longitud_pico (mm)", "profundidad_pico (mm)", "longitud_aleta (mm)", "masa_corporal (g)", "genero"])

df[20:30]
```

1¿Cuáles son las profundidades del pico máxima y mínima de
los pinguinos de la isla Torgersen?



```python
df_isla= df['isla']
df_prof= df['profundidad_pico (mm)']
dfprof2=[]
for i in range(len(df_isla)):
    if df_isla[i] == 'Torgersen':
        dfprof2.append(df_prof[i])
print('max:',max(dfprof2))
print('min:',min(dfprof2))
```

2. ¿De qué isla, especie y género es el pingüino con la mayor masa corporal? Ayuda: utilice la función argmax()


```python
dfmasa= df['masa_corporal (g)']
dfespecie= df['especie']
dfgenero= df['genero']

ind_mxm= dfmasa.argmax()
print('masa: ',dfmasa[ind_mxm], 'g')
print('especie: ',dfespecie[ind_mxm])
print('genero: ',dfgenero[ind_mxm])
```

3. Cuál es la masa corporal media de los pinguinos macho que no son de la especie Gentoo?

```python
masagento=[]
for i in range(len(dfespecie)):
  if dfespecie[i] != 'Gentoo' and dfgenero[i] == 'Male':
    masagento.append(dfmasa[i])

print(np.mean(masagento).round(2), 'g')
```

4. Grafique la distribución de la longitud del pico. Añada el valor medio y la desviación estándar en el título de la figura. Ayuda: utilice la funcion hist()

```python
df_pico = df['longitud_pico (mm)']
fig = plt.figure(figsize=(10, 6))
sns.histplot(df_pico)
#Notese que en la tarea se pide que se de la informacion en el titulo
plt.title(f'Distribucion de la longitud del pico \n media: {df_pico.mean().round(2)}, std {df_pico.std().round(2)}')

plt.ylabel('conteos')
plt.xlabel('Longitud del pico(mm)')
plt.grid()
plt.show()
```


5. ¿Cuáles son las islas y especies presentes en el conjunto de datos? ¿Con qué frecuencia aparecen? Grafique la distribución de la masa corporal para los pingüinos.
