````python
df_masa_corporal = df.loc[df["masa_corporal (g)"] == 6300.0,["especie","isla","genero"]]
````

````python
print (df_masa_corporal, "Pingüino con mayor masa corporal")
````

````python
df["masa_corporal (g)"].dtype
````

````python
df["masa_corporal (g)"] = pd.to_numeric(
    df["masa_corporal (g)"],
    errors="coerce"
)
````

````python
df["masa_corporal (g)"].argmax()
````

````python
filtro = (df["genero"] == "Male") & (df["especie"] != "Gentoo")

print (df.loc[filtro, "masa_corporal (g)"].mean(), "masa corporal media de los pinguinos macho que no son de la especie Gentoo")
````

````python
media = df["longitud_pico (mm)"].mean()
desviacion = df["longitud_pico (mm)"].std()

plt.hist(df["longitud_pico (mm)"], bins=20)

plt.xlabel("Longitud del pico (mm)")
plt.ylabel("Frecuencia")
plt.title(
    f"Distribución de la longitud del pico\n"
    f"Media = {media:.2f} mm | Desv. estándar = {desviacion:.2f} mm"
)

plt.show()
print("Valor medio de la longitud de pico (mm)", media )
print("Desviación estandar de la longitud de pico (mm)", desviacion)
````
