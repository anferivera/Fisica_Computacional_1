```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = sns.load_dataset("penguins").dropna().values
df = pd.DataFrame(data, columns=["especie", "isla", "longitud_pico (mm)", "profundidad_pico (mm)", "longitud_aleta (mm)", "masa_corporal (g)", "genero"])

df[:6]

```

```python

df_torgesen=df[df.isla=="Torgersen"]
df_torgesen[:6]
```

```python
df_torgesen.columns
```

```python
#Punto 1
min_T=df_torgesen['profundidad_pico (mm)'].min()
max_T=df_torgesen['profundidad_pico (mm)'].max()
print(f"Profundidad de pico mínima en Torgersen = {min_T} mm")
print(f"Profundidad de pico máxima en Torgersen = {max_T} mm")
```

```python
#Punto 2
indice = df['masa_corporal (g)'].argmax()
df.loc[indice, ['isla', 'especie', 'genero',"masa_corporal (g)"]]
```

```python
df_no_gentoo=df[df['especie'] != 'Gentoo']
df_no_gentoo[:6]
```

```python
#Punto 3
media = df_no_gentoo['masa_corporal (g)'].mean()
print("La media la masa corporal de los pinguinos macho que no son de la especie Gentoo")
print(media)
```

```python
#Punto 4
media = df['longitud_pico (mm)'].mean()
desviacion = df['longitud_pico (mm)'].std()
plt.hist(df['longitud_pico (mm)'])
plt.title(f'Longitud del pico: media = {media:.2f} mm, desviación estándar = {desviacion:.2f} mm')
plt.xlabel('Longitud del pico (mm)')
plt.ylabel('Frecuencia')
```

```python
#Punto 4: Islas
print(df['isla'].value_counts())
```

```python
#Punto 4: Especies
print(df['especie'].value_counts())
```

```python
#Punto 4: Distribución de la masa corporal
plt.hist(df['masa_corporal (g)'])
plt.xlabel('Masa corporal (g)')
plt.ylabel('Frecuencia')
plt.title('Distribución de la masa corporal de los pingüinos')
plt.show()
```
