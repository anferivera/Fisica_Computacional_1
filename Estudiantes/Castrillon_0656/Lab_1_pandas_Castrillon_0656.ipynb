````python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = sns.load_dataset("penguins").dropna().values
df = pd.DataFrame(data, columns=["especie", "isla", "longitud_pico (mm)", "profundidad_pico (mm)", "longitud_aleta (mm)", "masa_corporal (g)", "genero"])

df[:5]
````

```
[especie, isla, longitud del pico, profundidad del pico, longitud de la aleta, masa corporal, genero]
```

## Laboratorio 1

1.   ¿Cuáles son las profundidades del pico máxima y mínima de los pinguinos de la isla Torgersen?

````python
prof_pico_torgersen_max, prof_pico_torgersen_min = df[df["isla"] == "Torgersen"]["profundidad_pico (mm)"].max(), \
                                                   df[df["isla"] == "Torgersen"]["profundidad_pico (mm)"].min()
print(f"Profundidad del pico (mm) maxima: {prof_pico_torgersen_max}")
print(f"Profundidad del pico (mm) minima: {prof_pico_torgersen_min}")
````

2.    ¿De qué isla, especie y género es el pingüino con la mayor masa corporal?
       **Ayuda:** utilice la función `argmax()`

````python
solucion2 = df.iloc[[np.argmax(df["masa_corporal (g)"])]][["isla", "genero", "especie"]]
print(f"La isla, especie y género del pinguino con mayor masa corporal ({df["masa_corporal (g)"].max()} gramos) es: ")
display(solucion2)
````

3.   ¿Cuál es la masa corporal media de los pinguinos macho que no son de la especie Gentoo?

````python
print(f"Comprobación de que solamente hay dos géneros: {list(df["genero"].unique())}")

solucion3 = df[(df["genero"] == "Male") & (df["especie"] != "Gentoo")]["masa_corporal (g)"].mean()
print(f"El promedio de masa corporal de los pinguinos macho no Gentoo es: {solucion3.round(2)} gramos")
````

4.   Grafique la distribución de la longitud del pico. Añada el valor medio y la desviación estándar en el título de la figura. **Ayuda:** utilice la funcion `hist()`

````python
media = df["longitud_pico (mm)"].mean()
desv = df["longitud_pico (mm)"].std()

plt.hist(df["longitud_pico (mm)"], bins=20, edgecolor="black", color="skyblue")
plt.title(f"4. Distribución de la longitud del pico\nMedia = {media:.2f}, Desv. Est. = {desv:.2f}")
plt.xlabel("Longitud del pico (mm)")
plt.ylabel("Frecuencia")
plt.show()
````

5.   ¿Cuáles son las islas y especies presentes en el conjunto de datos? ¿Con qué frecuencia aparecen? Grafique la distribución de la masa corporal para los pingüinos.

````python
# Islas presentes y su frecuencia
print("Frecuencia de islas:")
print(df["isla"].value_counts(), "\n")

# Especies presentes y su frecuencia
print("Frecuencia de especies:")
print(df["especie"].value_counts(), "\n")

# Distribución de la masa corporal
plt.figure(figsize=(8,6))
plt.hist(df["masa_corporal (g)"], bins=20, edgecolor="black", color="skyblue")
plt.title("Distribución de la masa corporal de los pingüinos")
plt.xlabel("Masa corporal (g)")
plt.ylabel("Frecuencia")
plt.show()
````
