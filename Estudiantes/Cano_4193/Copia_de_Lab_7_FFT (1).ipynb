```python
import librosa as lb #libreria para ver los datos de manera dumerica de un audio
import scipy as sc
import numpy as np
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))
datos, fs = lb.load("do_mayor.wav", sr=None)
sf = sc.fft.fft(datos)
nf = np.fft.fft(datos)
ax1.set_title("Señal original")
ax1.plot(datos)
ax2.set_title("Transformada de Fourier")
ax2.plot(np.abs(sf))
ax2.set_xlim(0,1000)
```

```python
tiempo = np.arange(len(datos))/fs #1/fs = delta de t cada intervalo separado, segun la cantidad de datos
tiempo
```

```python
import numpy as np #codigo del cuaderno


def FT(x, y):
    N = len(x)
    dx = x[1] - x[0]  # Paso de muestreo
    f = np.fft.fftfreq(N, dx)  # Arreglo de frecuencias

    F = np.zeros(N, dtype=complex)
    for i, fi in enumerate(f):
        F[i] = np.sum(y * np.exp(-2j * np.pi * fi * x)) * dx

    return f, F
cosas = FT(tiempo,datos)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))
ax1.set_title("Señal orginial")
ax1.plot(datos)
ax2.set_title("Señal del codigo del profe reconstruida")
ax2.plot(sc.fft.ifft(cosas[1]))
plt.show()
```
