```python
df_limpio = df.dropna(subset=["mpg", "horsepower"])
sns.regplot(data = df_limpio, x="mpg",order=1, y= "horsepower",line_kws={"color": "red"})
print("el intervalo de confienza es de 95%")
```

```python
b, a = np.polyfit(df_limpio["mpg"], df_limpio["horsepower"], 1)
r = df["mpg"].corr(df["horsepower"])
```

```python
print("Pendiente =", b)
print("Intercepto =", a)
print("El coeficiente de correlacion es cercano a -1 por lo que se denota una fuerte correlación inversa",r)
print("El porcentaje de determinación es de:",r**2)
```

```python
url2 = "https://raw.githubusercontent.com/anferivera/SDdiracDM/master/general-scan.csv"
df2 = pd.read_csv(url2)
df2
```

```python
df2["Omega"]
df5 = df2[df2["Omega"] != 0.12]
df5["Omega"]
```

```python
from numpy._core.multiarray import MAY_SHARE_EXACT
M1 = df2["MDF"]
Mn = df2["YRC"]*df2["vS"]/2**(1/2)
df2["Mn"] = Mn
df3 = df2[df2["Omega"] == 0.12]
df5 = df2[df2["Omega"] != 0.12]
```

```python
df5["Omega"]
```

```python
df5["MDF"]
df5["Mn"]

good = df5

xmin, xmax = 1e2, 1e4
ymin, ymax = 1e1, 1e4

intervalo = np.logspace(np.log10(xmin), np.log10(xmax), 40)
centro, peque, grande = [], [], []
for i in range(len(intervalo) - 1):
    intervalo_MDF = (good["MDF"] >= intervalo[i]) & (good["MDF"] < intervalo[i + 1])
    if intervalo_MDF.sum() >= 2:
        centro.append(np.sqrt(intervalo[i] * intervalo[i + 1]))
        peque.append(good.loc[intervalo_MDF, "Mn"].min())
        grande.append(good.loc[intervalo_MDF, "Mn"].max())
centro, lo, grande = map(np.array, (centro, peque, grande))

x_env =  np.concatenate(([xmin], centro, [xmax]))
lo_env = np.concatenate(([peque[0]], peque, [peque[-1]]))
grande_env = np.concatenate(([grande[0]], grande, [grande[-1]]))

sns.set_theme(style="white")
fig, ax = plt.subplots(figsize=(6.5, 6))

ax.fill_between([xmin, xmax], ymin, ymax, color="red",alpha=0.5, zorder=0,
                 label="Under and over-abundance")
ax.fill_between(x_env, lo_env, grande_env, color="white", zorder=1)


zona_ver = 60
ax.fill_between([xmin, xmax], ymin, zona_ver, color="green", zorder=2,label="Excluded by DD exp.")

ax.scatter(good["MDF"], good["Mn"], s=6, color="blue", zorder=3,label=r"$\Omega h^2 - m_{\nu_\alpha}$")

ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlim(xmin, xmax)
ax.set_ylim(ymin, ymax)
ax.set_xlabel(r"$M_\Psi\,(\mathrm{GeV})$", fontsize=13)
ax.set_ylabel(r"$Mn\,(\mathrm{GeV})$", fontsize=13)

ax.text(2.5e2, 2.5e3, r"$\Omega_{\chi_1^0} < \Omega_{DM}$", fontsize=11)
ax.text(2e3, 4e3, r"$\Omega_{\chi_1^0} > \Omega_{DM}$", fontsize=11)

ax.legend(loc="lower right", frameon=True, fontsize=9)
plt.tight_layout()
plt.savefig("relic_density_plot.png", dpi=200)
plt.show()
```
