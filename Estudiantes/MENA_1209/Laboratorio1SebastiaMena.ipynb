```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


data = sns.load_dataset("penguins").dropna().values
df = pd.DataFrame(data, columns=["especie", "isla", "longitud_pico (mm)", "profundidad_pico (mm)", "longitud_aleta (mm)", "masa_corporal (g)", "genero"])

df[:5]
```

```python
# Convertir columnas numéricas a float --- para evitar problemas de incompatibilidad entre el tipo de dato.
cols_num = ["longitud_pico (mm)", "profundidad_pico (mm)", "longitud_aleta (mm)", "masa_corporal (g)"]
df[cols_num] = df[cols_num].astype(float)
```

## 1. ¿Cuáles son las profundidades del pico máxima y mínima de los pinguinos de la isla Torgersen?

```python
## profundidades del pico máxima y mínima de los pinguinos de la isla Torgersen

# Filtrar por isla
torgersen = df[df["isla"] == "Torgersen"]

# Calcular máximo y mínimo
prof_max = torgersen["profundidad_pico (mm)"].max()
prof_min = torgersen["profundidad_pico (mm)"].min()

print(f"Profundidad máxima: {prof_max} mm")
print(f"Profundidad mínima: {prof_min} mm")
```

## 2. ¿De qué isla, especie y género es el pingüino con la mayor masa corporal? Ayuda: utilice la función argmax()

```python
## Encontrar el índice de la masa corporal máxima
idx_max = df["masa_corporal (g)"].argmax()

# Obtener los datos de esa fila específica
pinguino_max = df.iloc[idx_max]

print(f"Isla: {pinguino_max['isla']}")
print(f"Especie: {pinguino_max['especie']}")
print(f"Género: {pinguino_max['genero']}")
```

## 3. ¿Cuál es la masa corporal media de los pinguinos macho que no son de la especie Gentoo?

```python
# Crear el filtro lógico
mascara = (df["genero"] == "Male") & (df["especie"] != "Gentoo")

# Aplicar el filtro y calcular la media
media_masa = df[mascara]["masa_corporal (g)"].mean()

print(f"Masa corporal media: {media_masa:.2f} g")
```

## 4. Grafique la distribución de la longitud del pico. Añada el valor medio y la desviación estándar en el título de la figura. Ayuda: utilice la funcion hist()

```python


# Extraer los datos y calcular métricas
longitudes = df["longitud_pico (mm)"]
media = longitudes.mean()
desviacion = longitudes.std()

# Crear el histograma
plt.hist(longitudes, bins=15, color='skyblue', edgecolor='black')

# Añadir título con variables
plt.title(f"Distribución de la Longitud del Pico\nMedia: {media:.2f} mm | Desviación: {desviacion:.2f} mm")
plt.xlabel("Longitud del pico (mm)")
plt.ylabel("Frecuencia")

# Mostrar el gráfico
plt.show()
```

## 5. ¿Cuáles son las islas y especies presentes en el conjunto de datos? ¿Con qué frecuencia aparecen? Grafique la distribución de la masa corporal para los pingüinos.

```python
# Mostrar frecuencias
print("Frecuencia por Islas:")
print(df["isla"].value_counts())

print("\nFrecuencia por Especies:")
print(df["especie"].value_counts())

# Graficar la distribución de la masa corporal
plt.figure() # Crea una nueva figura
plt.hist(df["masa_corporal (g)"], bins=15, color='lightgreen', edgecolor='black')
plt.title("Distribución de la Masa Corporal de los Pingüinos")
plt.xlabel("Masa corporal (g)")
plt.ylabel("Frecuencia")
plt.show()
```
