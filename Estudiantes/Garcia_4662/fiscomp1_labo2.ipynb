```python
## 1
cosa = pd.read_csv("millas_por_galon.csv")
df = pd.DataFrame(cosa, columns=["horsepower", "mpg"])

plt.figure(figsize=(10, 6))
sns.regplot(x="horsepower", y="mpg", data=df, ci=95, line_kws={"color":"red"})
plt.title("Millas por Galón (mpg) vs Potencia (horsepower) al 95%")
plt.xlabel("Potencia (horsepower)")
plt.ylabel("Millas por Galón (mpg)")
plt.grid(True)
plt.show()
```
