```python
data = sns.load_dataset("penguins").dropna().values
torgersen = pd.DataFrame(data, columns=["especie", "isla", "longitud_pico (mm)", "profundidad_pico (mm)", "longitud_aleta (mm)", "masa_corporal (g)", "genero"])
torgersen
```

```python
## 1.

print("Profundidad de pico máxima: ", torgersen["profundidad_pico (mm)"].astype(float).max(), "(mm)")
print("Profundidad de pico mínima: ", torgersen["profundidad_pico (mm)"].astype(float).min(), "(mm)")
```

```python
## 2

pingu_g_arg = torgersen["masa_corporal (g)"].astype(float).argmax()
isla_ = torgersen["isla"][pingu_g_arg]
especie_ = torgersen["especie"][pingu_g_arg]
genero_ = torgersen["genero"][pingu_g_arg]

print(r"Del pinguino más gordo: ")
print(f"Isla: {isla_}")
print(f"Especie: {especie_}")
print(f"Género: {genero_}")
```

```python
## 3.

lista_p = torgersen.loc[(torgersen.genero == "Male") & (torgersen.especie != "Gentoo"), :]
masa_media = lista_p["masa_corporal (g)"].mean()
print("masa media =",masa_media.round(3))
```

```python
## 4.

mu = torgersen["longitud_pico (mm)"].astype(float).mean()
sigma = torgersen["longitud_pico (mm)"].astype(float).std()

plt.hist(torgersen["longitud_pico (mm)"], bins=15, edgecolor="k")
plt.title(fr"Distribución de longitudes de pico; $\mu =$ {mu.round(3)} , $\sigma =$ {5.469}")
plt.xlabel("Longitud de pico (mm)")
plt.ylabel("Frecuencia")
plt.grid()
plt.show()
```

```python
## 5.

## Frecuencia y nombre de las islas
print("Frecuencia de las islas")
print(torgersen["isla"].value_counts())

## Frecuencia y nombre de las especies
print("Frecuencia de las especies")
print(torgersen["especie"].value_counts())

## Dsitribución de masa
mu_m = torgersen["masa_corporal (g)"].astype(float).mean()
sigma_m = torgersen["masa_corporal (g)"].astype(float).std()

#print(mu_m) 4207.057057
#print(sigma_m) 805.2158019

plt.hist(torgersen["masa_corporal (g)"], bins=15, edgecolor="k")
plt.title(fr"Distribución de longitudes de pico; $\mu =$ {4207.06} , $\sigma ={805.21}$ ")
plt.xlabel("Masa corporal (g)")
plt.ylabel("Frecuencia")
plt.grid()
plt.show()
```
