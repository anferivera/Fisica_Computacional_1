```python
#Inciso 2

url = "https://raw.githubusercontent.com/anferivera/SDdiracDM/master/general-scan.csv"

df = pd.read_csv(url)
df.head()
```

```python
df = df.drop(columns=["Unnamed: 0"])
df
```

```python
#Cálculo de las variables para la gráfica

df["MN"] = df["YRC"] * df["vS"] / np.sqrt(2) 
df["MPsi"] = df["MDF"] 

#Valor observado de la densidad de materia oscura
omega_observada = 0.12
tol = 0.01

omega_min = omega_observada-tol
omega_max = omega_observada+tol

# Puntos con la abundancia correcta
correctos = df[(df["Omega"] >= omega_min) & (df["Omega"] <= omega_max)] #Puntos con la abundancia correcta


#Construcción de la gráfica
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlim(100, 10000)
ax.set_ylim(10, 10000)

#Zona roja superior
xr_left = np.array([100, 1000])
yr_left = np.array([100, 1000])
ax.fill_between(xr_left, yr_left, 10000, color='salmon', zorder=1) #para rellenar la región entre las curvas

xr_right = np.array([1300, 3000, 5000, 8000, 10000])
yr_right = np.array([1500, 2200, 2000, 1900, 1100])
ax.fill_between(xr_right, yr_right, 10000, color='salmon', zorder=1) 

#Zona verde
ax.axhspan( 10, 60, color="#42a052", zorder=1) #realiza la franja verde inferior
ax.axvspan( 900, 1300, ymin=0.55, ymax=1.0, color="#42a052", zorder=1)  #empieza al 55 % de la altura de la gráfica y termina en el 100 % de la altura.

#puntos azules
ax.scatter( correctos['MPsi'], correctos['MN'], color='blue', s=9, label=r'$\Omega h^2 - m_{\nu_\alpha}$', zorder=3,)

#Complementos de texto
ax.text(150, 900, r'$\mathbf{\Omega_{\chi_1^0} < \Omega_{DM}}$', fontsize=18)
ax.text(2200, 2500, r'$\mathbf{\Omega_{\chi_1^0} > \Omega_{DM}}$', fontsize=18)
ax.set_xlabel(r'$M_{\Psi}\ \mathbf{(GeV)}$', fontsize=12)
ax.set_ylabel(r'$M_N\ \mathbf{(GeV)}$', fontsize=12)
blue_dots = plt.Line2D( [0], [0], marker='o', color='w', markerfacecolor='blue', markersize=5, label=r'$\Omega h^2 - m_{\nu_\alpha}$',)
ax.legend(loc='lower right', frameon=True, facecolor='#eAF4eA', edgecolor='gray', fontsize=10,)
plt.tight_layout()
plt.show()
```
