```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

## Laboratorio 2
* Hacer el analisis estadístico de las variables: Millas por galón (mpg) vs potencia (horsepower) in the dataset: millas_por_galon.csv. Contruya el fit de al $95$ C.L.     



  



```python
#Importamos los datos
ruta = r'/content/millas_por_galon (1).csv'
df = pd.read_csv (ruta)

df
```

```python
#Analisis estadistico
Datos= df[['mpg','horsepower']].describe()
print(Datos)
```

```python
#Graficas
sns.set_theme(context="notebook")

sns.relplot(
    data=df,
    x='mpg',y='horsepower',
    height=4,aspect= 2,
    s=60,              # tamaño de los puntos
    color='darkblue',# color (si no usas hue)
    alpha=0.6,         # transparencia
    marker='D',        # forma: 'o', 's', 'D', '^', 'X', etc.
    edgecolor='magenta', # color del borde
    linewidth=0.5,     # grosor del borde

)
plt.title('Comparativa caballos de fuerza VS millas por galon.')
```

```python
# Regresión polinomial con intervalo de confianza
sns.regplot(
    data=df,
    x='mpg',y='horsepower',
    order=4,  # Grado del polinomio
    ci=95,
    scatter_kws={
        'color': 'blue',
        's': 60,          # tamaño
        'alpha': 0.5,      # transparencia
        'edgecolor': 'black',
        'linewidths': 0.5
    },


    line_kws={'color':'magenta'},
)

plt.title('Relación MPG vs. Horsepower (Ajuste Polinomial + 95% C.L.)')
plt.show()
```

  desde GitHUb y reconstruir la figura 6 del [trabajo](https://journals.aps.org/prd/abstract/10.1103/PhysRevD.100.035029). Para ello usar `seaborn` y `matplolib`.

  
  * $M_N=h_s\dfrac{v_s}{\sqrt{2}}$
  * $v_s =$ vS
  * $h_s=$ YRC
  * $M_\Psi=$ MDF  
  * $M_{\chi_1^0}=$ mChi1
  * $\Omega_{\chi_1^0}=$ Omega
  * $\Omega_{DM}=0.12$

```python
url = 'https://raw.githubusercontent.com/anferivera/SDdiracDM/master/general-scan.csv'
datos_2 = pd.read_csv(url)
datos_2
```

```python
#Creamos un vector con los datos para graficar
m_n = datos_2['YRC'] * datos_2['vS']/ np.sqrt(2)

fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlim(100, 10000)
ax.set_ylim(5, 10000)

plt.scatter(x=datos_2['MDF'],y=m_n, color ='magenta')
plt.show()
```

```python
datos_2.columns
```

```python
datos_2['M_N'] = datos_2['YRC'] * (datos_2['vS'] / np.sqrt(2))
datos_2['M_Psi'] = datos_2['MDF']

Omega_DM = 0.12
tol = 0.01
df_correctos = datos_2[
    (datos_2['Omega'] >= Omega_DM - tol) & (datos_2['Omega'] <= Omega_DM + tol)
]

# 3. Configuración de las graficas
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlim(100, 10000)
ax.set_ylim(8, 10000)


x_red_left = np.array([100, 1000])
y_red_left = np.array([100, 1000])
ax.fill_between(
    x_red_left, y_red_left, 10000, color='#fa8072', alpha=0.9, zorder=1
)

x_red_right = np.array([1300, 3000, 5000, 8000, 10000])
y_red_right = np.array([1500, 2000, 1800, 1500, 1200])
ax.fill_between(
    x_red_right, y_red_right, 10000, color='#fa8072', alpha=0.9, zorder=1
)
ax.fill_between(
    [1000, 10000], [10000, 10000], [10000, 10000], color='#fa8072', zorder=1
)


ax.axhspan(
    8, 60, color='#4ba359', zorder=1
)
ax.axvspan(
    900, 1300, ymin=0.55, ymax=1.0, color='#4ba359', zorder=1
)

ax.scatter(
    df_correctos['M_Psi'],
    df_correctos['M_N'],
    color='blue',
    s=6,
    alpha=0.8,
    label=r'$\Omega h^2 - m_{\nu_\alpha}$',
    zorder=3,
)

ax.text(
    150, 900, r'$\mathbf{\Omega_{\chi_1^0} < \Omega_{DM}}$', fontsize=13
)
ax.text(
    2200, 2500, r'$\mathbf{\Omega_{\chi_1^0} > \Omega_{DM}}$', fontsize=13
)


ax.set_xlabel(r'$M_{\Psi}\ \mathbf{(GeV)}$', fontsize=12)
ax.set_ylabel(r'$M_N\ \mathbf{(GeV)}$', fontsize=12)

blue_dots = plt.Line2D(
    [0],
    [0],
    marker='o',
    color='w',
    markerfacecolor='blue',
    markersize=5,
    label=r'$\Omega h^2 - m_{\nu_\alpha}$',
)

ax.legend(
    loc='lower right',
    frameon=True,
    facecolor='#eAF4eA',
    edgecolor='gray',
    fontsize=10,
)

plt.tight_layout()
plt.show()
```
