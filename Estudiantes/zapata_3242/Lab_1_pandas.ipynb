1.   ¿Cuáles son las profundidades del pico máxima y mínima de los pinguinos de la isla Torgersen?

````python
#dataframe de solo la isla Torgersen
df_Torgersen = df[df.isla== 'Torgersen']
df_Torgersen    
````

````python
print('Profundidad Minima=',df_Torgersen.min())
print('Profundidad Maxima=',df_Torgersen.max())
````

````python
print('profundidades minima del pico de los pinguinos de la isla Torgersen=',df_Torgersen['profundidad_pico (mm)'].min(),'mm')
print('profundidades maxima del pico de los pinguinos de la isla Torgersen=',df_Torgersen['profundidad_pico (mm)'].max(), 'mm')
````

2.    ¿De qué isla, especie y género es el pingüino con la mayor masa corporal?

````python
masa_max = np.argmax(df['masa_corporal (g)'])
info_masa_max = df.iloc[masa_max]
info_masa_max
````

````python
print('la isla, especie y género del pingüino con la mayor masa corporal es:')
print('isla =', info_masa_max['isla'])
print('especie =', info_masa_max['especie'])
print('genero =', info_masa_max['genero'])
````

3.   ¿Cuál es la masa corporal media de los pinguinos macho que no son de la especie Gentoo?


````python
df_nogentoo = df[np.logical_and(np.logical_or(df.especie == 'Adelie', df.especie =='Chinstrap'),df.genero == 'Male')]
df_nogentoo
````

````python
mean_nongetoo = np.mean(df_nogentoo['masa_corporal (g)'])
print(f'Media masa corporal de los pinguinos que no son especie Gentoo ={mean_nongetoo : .3f} g ')
````

Pregunta 4

````python
sns.histplot(data= df, x='longitud_pico (mm)')

medio =  np.mean(df['longitud_pico (mm)'])
std = np.std(df['longitud_pico (mm)'])
 
plt.title(f'distribución de la longitud del pico \nvalor medio {medio: .2f} y desviación estándar {std: .2f}')
plt.show()
````
